```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

sns.set_style("whitegrid")
```

```python
PROJECT_ROOT = Path.cwd().resolve().parent
DATA_PATH = PROJECT_ROOT / "data" / "curated_dataset_summary.csv"
df = pd.read_csv(DATA_PATH)
df = df[df["n_days"] != 1].copy()
```

```python
fig, ax = plt.subplots(figsize=(14, 10), nrows=2, ncols=2)
ax = ax.flatten()

study_vars = ["n_participants", "n_days", "n_beeps", "n_rows"]
titles = ["Number of Participants", "Number of Days", "Number of Beeps per Day", "Number of Observations"]
colors = sns.color_palette(n_colors=4)

for i, col in enumerate(study_vars):
    sns.violinplot(
        data=df,
        y=col,
        ax=ax[i],
        color=colors[i],
        inner="point",
        cut=2
    )

    ax[i].set_title(titles[i], fontsize=12)
    ax[i].set_xlabel("")
    ax[i].set_ylabel("")

fig.suptitle(
    "Distribution of Curated Dataset Characteristics",
    fontsize=18,
    fontweight="bold",
)

plt.tight_layout()
plt.show()
```
